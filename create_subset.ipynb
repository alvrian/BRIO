```python
import shutil
import os

path = 'cnndm/diverse/train/'
dest = './cnndm_subset/diverse/'
filename = os.listdir(path)

number_of_files = len(filename)
print(number_of_files)
```

```python
subset_size = 20000
train_size = 0.7
test_size = 0.2
valid_size = 0.1

for i in range(subset_size):
    if i < subset_size * train_size:
        dest = './cnndm_subset/diverse/train/'
    elif i < subset_size * (train_size + test_size):
        dest = './cnndm_subset/diverse/test/'
    else:
        dest = './cnndm_subset/diverse/val/'
    f = filename[i]
    if f.endswith('.json'):
        shutil.copy(os.path.join(path, f), dest)
        
print('summary of the number of files in each subset:')
print('Train:', len(os.listdir('./cnndm_subset/diverse/train/')))
print('Test:', len(os.listdir('./cnndm_subset/diverse/test/')))
print('Validation:', len(os.listdir('./cnndm_subset/diverse/val/')))
```
