## 5. Training

BRIO training uses a **dual loss**:
- **MLE Loss**: Standard cross-entropy on the gold summary (generation quality)
- **Ranking Loss**: Contrastive margin loss ensuring the model assigns higher probability to better candidates

### Key Hyperparameters (CNNDM defaults)
| Parameter | Value | Description |
|-----------|-------|-------------|
| `batch_size` | 1 | Per-GPU batch size |
| `accumulate_step` | 8 | Gradient accumulation steps |
| `mle_weight` | 0.1 | Weight for MLE loss |
| `rank_weight` | 10 | Weight for ranking loss |
| `max_lr` | 2e-3 | Peak learning rate |
| `warmup_steps` | 10000 | LR warmup steps |
| `eval_interval` | 1000 | Steps between evaluations |
| `max_num` | 16 | Number of candidate summaries |
| `model_type` | `facebook/bart-large-cnn` | Base model |

> **⚠️ Training Time**: Full CNNDM training (287K samples) takes **many hours** even on a V100/A100. For a quick sanity check, reduce `eval_interval` or use a small data subset.

### 6c. Calculate ROUGE Scores
Compute ROUGE-1, ROUGE-2, and ROUGE-L scores using the Python implementation.

````python
# ROUGE for generator output (using Python implementation)
# -p flag uses compare-mt's Python ROUGE (no Perl ROUGE setup needed)
# -l flag lowercases the text before scoring

# First, create tokenized version using NLTK (simpler alternative to Stanford CoreNLP)
from nltk import word_tokenize, sent_tokenize

def tokenize_file(input_path, output_path):
    """Tokenize a file using NLTK word tokenizer."""
    with open(input_path) as fin, open(output_path, 'w') as fout:
        for line in fin:
            tokens = ' '.join(word_tokenize(line.strip()))
            fout.write(tokens + '\n')
    print(f'Tokenized: {input_path} -> {output_path}')

# Tokenize generated output and reference
if os.path.exists('./result/cnndm/test.out'):
    tokenize_file('./result/cnndm/test.out', './result/cnndm/test.out.tokenized')

if os.path.exists('./cnndm/test.target') and not os.path.exists('./cnndm/test.target.tokenized'):
    tokenize_file('./cnndm/test.target', './cnndm/test.target.tokenized')
````

## 7. Quick Demo: Use Pretrained BRIO from HuggingFace

No data download or training needed — just load the pretrained model and generate summaries.

Available models:
- `Yale-LILY/brio-cnndm-uncased` (BART, CNNDM, lowercase)
- `Yale-LILY/brio-cnndm-cased` (BART, CNNDM, cased)
- `Yale-LILY/brio-xsum-cased` (Pegasus, XSum, cased)

````python
from transformers import BartTokenizer, BartForConditionalGeneration

# Load pretrained BRIO (CNNDM, uncased)
model_name = 'Yale-LILY/brio-cnndm-uncased'
tokenizer = BartTokenizer.from_pretrained(model_name)
model = BartForConditionalGeneration.from_pretrained(model_name)
model.eval()

print(f'Model loaded: {model_name}')
print(f'Parameters: {sum(p.numel() for p in model.parameters()) / 1e6:.1f}M')
````

````python
# Summarize a sample article
article = """
Manchester United superstar Cristiano Ronaldo scored his 806th career goal in Old Trafford, 
breaking FIFA's all-time record for most goals in competitive matches in men's football history. 
It was the second of three goals the Portuguese attacker scored during the game, 
leading United to a 3-2 victory over Tottenham and finishing the day with 807 total career goals. 
The previous FIFA goal record was held by Josef Bican, with 805 goals.
""".strip().lower()  # lowercase for uncased model

inputs = tokenizer([article], max_length=1024, return_tensors='pt', truncation=True)

import torch
with torch.no_grad():
    summary_ids = model.generate(
        inputs['input_ids'],
        max_length=140,
        min_length=55,
        num_beams=4,
        length_penalty=2.0,
        no_repeat_ngram_size=3,
        early_stopping=True,
    )

summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True, clean_up_tokenization_spaces=False)
print('=== Article ===')
print(article)
print('\n=== Generated Summary ===')
print(summary)
````
